## Understanding K-Means Clustering

K-means clustering is a popular unsupervised machine learning algorithm used for partitioning a dataset into distinct, non-overlapping subsets or clusters. The goal is to group similar data points together and discover underlying patterns in the data.
For more information check: [K-means_on wikipedia](https://en.wikipedia.org/wiki/K-means_clustering)

K-means clustering finds applications in various fields, including customer segmentation, image compression, and anomaly detection, making it a fundamental tool in data analysis and pattern recognition.

### How K-Means Works:

1. **Initialization:**
   - Choose the number of clusters (\(k\)) you want to identify in the dataset.
   - Randomly place \(k\) centroids in the feature space, representing the initial cluster centers.


2. **Assigning Data Points:**
   - Calculate the distance between each data point and the centroids.
   - Assign each data point to the nearest centroid, forming \(k\) clusters.


3. **Updating Centroids:**
   - Recalculate the centroids of the clusters by finding the mean of all data points assigned to each cluster.
   - The centroids represent the new cluster centers.


4. **Iterative Refinement:**
   - Repeat the assignment and centroid update steps until convergence.
   - Convergence occurs when the centroids no longer change significantly or after a predefined number of iterations.


5. **Result:**
   - The final centroids represent the centers of the clusters.
   - Each data point belongs to the cluster whose centroid is nearest.

### Key Considerations:
- K-means is sensitive to the initial placement of centroids. Different initializations can lead to different clusterings.
- It is essential to choose an appropriate value for \(k\). 
- The algorithm converges to a local optimum, so multiple runs with different initializations are recommended to obtain a more stable solution.

### Step 0: Visualizing Predefined Clusters

#### Exercise Description:
1. Create four clusters of points on a 2D plane.
2. Visualize these points using `matplotlib` to understand the initial setup.

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42) # we fix the seed so we get always the same "random" points

# Define the four clusters of points
cluster1 = np.random.normal(loc=[3, 7], scale=[1.5, 1.5], size=(20, 2))
cluster2 = np.random.normal(loc=[7, 5], scale=[1, 1], size=(20, 2))
cluster3 = np.random.normal(loc=[5, 10], scale=[2, 2], size=(30, 2))
cluster4 = np.random.normal(loc=[9, 8], scale=[1, 1], size=(20, 2))

# Plot the clusters
plt.figure(figsize=(8, 6))
plt.scatter(cluster1[:, 0], cluster1[:, 1], color='red', label='Cluster 1')
plt.scatter(cluster2[:, 0], cluster2[:, 1], color='blue', label='Cluster 2')
plt.scatter(cluster3[:, 0], cluster3[:, 1], color='green', label='Cluster 3')
plt.scatter(cluster4[:, 0], cluster4[:, 1], color='orange', label='Cluster 4')

# Set plot labels and title
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Predefined Clusters of Points')

# Show legend
plt.legend()

# Show the plot
plt.grid(True)
plt.show()
```

### Step 1: Initializing Centroids and Assigning Points

#### Exercise Description:
1. Initialize four centroids for the clusters of points you visualized in Step 0.
2. Assign each data point to the nearest centroid based on Euclidean distance.
3. Visualize the centroids and the points, each colored according to its assigned cluster.

```python
from sklearn.metrics import pairwise_distances_argmin_min

# Step 0: Define the clusters (reuse the cluster data from Step 0)
clusters = np.concatenate([cluster1, cluster2, cluster3, cluster4])

num_clusters = 4 # we suppose that we have 4 clusters

# Step 1: Initialize the centroids randomly
initial_centroids_indices = np.random.choice(clusters.shape[0], num_clusters, replace=False)
initial_centroids = clusters[initial_centroids_indices]


# Assign points to nearest centroid
labels, _ = pairwise_distances_argmin_min(clusters, initial_centroids)

# Extract points for each cluster based on labels
cluster1_points = clusters[labels == 0]
cluster2_points = clusters[labels == 1]
cluster3_points = clusters[labels == 2]
cluster4_points = clusters[labels == 3]
```

```python
# we define a plot_cluster function that we will call after each step to check the modifications
import matplotlib.pyplot as plt

def plot_clusters(data, labels, centroids, num_clusters):
    plt.figure(figsize=(8, 6))
    colors = ['red', 'blue', 'green', 'orange', 'cyan']  # You can extend this list for more colors
    for i in range(num_clusters):
        cluster_points = data[labels == i]
        plt.scatter(cluster_points[:, 0], cluster_points[:, 1], color=colors[i], label=f'Cluster {i + 1}')
    plt.scatter(centroids[:, 0], centroids[:, 1], color='black', marker='X', label='Centroids')
    plt.xlabel('X-axis')
    plt.ylabel('Y-axis')
    plt.title('Clusters with Centroids')
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
# Example usage of the plot_clusters function
plot_clusters(clusters, labels, initial_centroids, num_clusters)
```

In this function, `data` represents the data points, `labels` represents the cluster assignments for each data point, `centroids` represents the cluster centroids, and `num_clusters` represents the number of clusters. The function plots the clusters, assigns different colors to different clusters, plots the centroids, and displays the plot.

You can call this `plot_clusters` function after each step to visualize the clusters and centroids, making it easier to understand the algorithm's progress.

### Step 2: Update Cluster Centroids (Function)

```python
def update_centroids(data, labels, num_clusters):
    centroids = np.array([data[labels == j].mean(axis=0) for j in range(num_clusters)])
    return centroids
```

**Usage:**

```python
# Example usage of the update_centroids function
updated_centroids = update_centroids(clusters, labels, num_clusters)
```

In this function, `data` represents the data points, `labels` represents the cluster assignments for each data point, and `num_clusters` represents the number of clusters. The function calculates and returns the updated centroids based on the mean of the points in each cluster.

```python
# we check the update with our plot function
plot_clusters(clusters, labels, updated_centroids, num_clusters)
```

### Step 3: Assign Data Points to Nearest Centroid (Function)

```python
from sklearn.metrics import pairwise_distances_argmin_min

def assign_to_centroids(data, centroids):
    labels, _ = pairwise_distances_argmin_min(data, centroids)
    return labels
```

**Usage:**


```python
# Example usage of the assign_to_centroids function
final_labels = assign_to_centroids(clusters, updated_centroids)
```

In this function, `data` represents the data points, and `centroids` represents the cluster centroids. The function calculates and returns the cluster assignments for each data point based on the nearest centroid.

Using these separate functions allows for better modularity and organization of the code. You can use these functions within the Step 4 code or any other part of your program where these specific steps are needed.

```python
# we check the update with our plot function
plot_clusters(clusters, final_labels, updated_centroids, num_clusters)
```

### Step 4: Iterative Refinement (and Convergence)

```python
def kmeans_clustering(data, num_clusters, max_iterations, tolerance):
    # Step 1: Initialize centroids (randomly or by other methods)
    initial_centroids = data[np.random.choice(data.shape[0], num_clusters, replace=False)]
    
    for i in range(max_iterations):
        # Step 2: Assign data points to nearest centroid
        labels = assign_to_centroids(data, initial_centroids)
        
        # Step 3: Update centroids based on mean of cluster points
        updated_centroids = update_centroids(data, labels, num_clusters)
        
        # Plot progress
        plot_clusters(data, labels, updated_centroids, num_clusters)
        
        # Check convergence
        if np.linalg.norm(updated_centroids - initial_centroids) < tolerance:
            break
        
        initial_centroids = updated_centroids
    
    return labels, updated_centroids


```

**Usage:**

```python
# Example usage
#num_clusters = 4 # this is already defined above
max_iterations = 10 # we limit to 10
tolerance = 1e-4

# Perform k-means clustering
final_labels, final_centroids = kmeans_clustering(clusters, num_clusters, max_iterations, tolerance)

# Plot the final clusters and centroids
plot_clusters(clusters, final_labels, final_centroids, num_clusters)
```

### Step 5: Result

In this implementation, the `kmeans_clustering` function takes the data points (`data`), the desired number of clusters (`num_clusters`), the maximum number of iterations (`max_iterations`), and the convergence tolerance (`tolerance`) as input parameters. It initializes centroids, assigns points to centroids, updates centroids, and iteratively refines the centroids until convergence. Finally, it returns the cluster labels and the final centroids.

After performing the clustering, the `plot_clusters` function is called to visualize the final clusters and centroids. You can adjust the `num_clusters`, `max_iterations`, and `tolerance` values based on your specific requirements.
