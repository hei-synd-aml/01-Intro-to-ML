## 1.  Introduction to Jupyter Notebooks

Welcome to the world of Jupyter Notebooks! Jupyter Notebooks are interactive documents that allow you to create and share documents containing live code, equations, visualizations, and narrative text. They are widely used in data science, research, education, and more.

### How Jupyter Notebooks Work:

- **Cells:** Notebooks are made up of cells. There are two main types of cells: Code cells and Markdown cells.
  
  - **Code Cells:** These cells contain executable code in Python, or other programming languages. You can run a code cell and see the output immediately below the cell.
  
  - **Markdown Cells:** These cells (as the one you are reading right now) contain formatted text, images, links, and more. You can use Markdown cells to add explanations, headings, and multimedia content to your notebook.

- **Interactivity:** Notebooks allow you to interact with code. You can edit and re-run code cells, making it a perfect environment for learning and experimentation.

- **Rich Outputs:** Notebooks can display rich outputs, including interactive plots, tables, and multimedia elements. This makes them great for data analysis and visualization.

### Getting Started:

1. **Working with Cells:** Click on a cell to select it. You can edit code cells or Markdown cells by double-clicking. Use the "Run" button or press `Shift + Enter` to execute a cell.

3. **Markdown Formatting:** In Markdown cells, you can use simple syntax to format text. For example, `**bold**`, `*italic*`, and `[links](https://example.com)`.
If you need some help for formatting, check: [markdown cheat-sheet (simple)](https://www.markdownguide.org/cheat-sheet/) or [markdown cheat-sheet (advanced)](https://github.com/adam-p/markdown-here/wiki/Markdown-Cheatsheet).
