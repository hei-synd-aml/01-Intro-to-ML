### Exercise 4: Functions

1. Create (define) a function called `calculate_factorial` that takes an integer as input and returns its factorial. (Bonus: propose a recursive version of the function)
2. Create a function called `is_prime` that takes a number as input and returns `True` if it's prime, `False` otherwise.

````python
# calculate factorial (non recursive)
def calculate_factorial(n):
    # Initialize the factorial to 1
    factorial = 1

    # Use a for loop to calculate the factorial
    for i in range(1, n + 1):
        factorial *= i

    # Return the factorial
    return factorial

# calculate factorial (recursive)
def calculate_factorial_rec(n):
    # Base case: factorial of 0 is 1
    if n == 0:
        return 1
    # Recursive case: n! = n * (n-1)!
    else:
        return n * calculate_factorial_rec(n - 1)

````

````python
# call to the function (for testing)
print(calculate_factorial(5)) # should print 120
print(calculate_factorial_rec(6)) # should print 720
````

````python
# is prime?

def is_prime(n):
    # 0 and 1 are not prime numbers
    if n < 2:
        return False
    # 2 is the only even prime number
    if n == 2: 
        return True    
    # All other even numbers are not primes
    if n % 2 == 0: 
        return False
    # Range starts with 3 and only needs to go up the square root of n
    # for all odd numbers
    for x in range(3, int(n**0.5)+1, 2):
        if n % x == 0:
            return False
    return True
````

````python
# call to the function (for testing)
print(is_prime(7)) # should print True
print(is_prime(309)) # should print False
````
